```python
import pandas as pd
import numpy as np
```

```python
columns = [
'srcip','sport','dstip','dsport','proto','state','dur',
'sbytes','dbytes','sttl','dttl','sloss','dloss','service',
'Sload','Dload','Spkts','Dpkts','swin','dwin','stcpb','dtcpb',
'smeansz','dmeansz','trans_depth','res_bdy_len','Sjit','Djit',
'Stime','Ltime','Sintpkt','Dintpkt','tcprtt','synack','ackdat',
'is_sm_ips_ports','ct_state_ttl','ct_flw_http_mthd','is_ftp_login',
'ct_ftp_cmd','ct_srv_src','ct_srv_dst','ct_dst_ltm','ct_src_ltm',
'ct_src_dport_ltm','ct_dst_sport_ltm','ct_dst_src_ltm',
'attack_cat','Label'
]
```

```python
unsw_nb15_testing = pd.read_csv("UNSW_NB15_testing-set.csv")
unsw_nb15_training = pd.read_csv("UNSW_NB15_training-set.csv")
unsw_nb15_1 = pd.read_csv("UNSW-NB15_1.csv",names=columns,low_memory=False)
unsw_nb15_2 = pd.read_csv("UNSW-NB15_2.csv",names=columns,low_memory=False)
unsw_nb15_3 = pd.read_csv("UNSW-NB15_3.csv",names=columns,low_memory=False)
unsw_nb15_4 = pd.read_csv("UNSW-NB15_4.csv",names=columns,low_memory=False)
unsw_nb15_list_events = pd.read_csv("UNSW-NB15_LIST_EVENTS.csv")
```

```python
unsw_nb15 = pd.concat([unsw_nb15_1, unsw_nb15_2, unsw_nb15_3, unsw_nb15_4], axis=0)
unsw_nb15.reset_index(drop=True, inplace=True)
```

```python
pd.set_option('display.max_columns', 1000)
pd.set_option('display.max_rows', 1000)
pd.set_option('display.width', 1000)
pd.set_option('display.max_colwidth', 1000)
```

```python
dfs  = {
    "unsw_nb15_testing": unsw_nb15_testing,
    "unsw_nb15_training": unsw_nb15_training,
    "unsw_nb15": unsw_nb15,
    "unsw_nb15_list_events": unsw_nb15_list_events
}
for name, df in dfs.items():
    print(f"\n DataFrame: {name}----{df.shape}")
    print(f"\n {df.head(5)}")
    print(f"\n Columns: ")
    print(df.columns)
    print(f"================================================================================================================")
```

```python
data = unsw_nb15_list_events.groupby("Attack category")['Number of events'].sum().sort_values(ascending=False)
log_data = np.log1p(data)
log_data.to_excel("attack_category_events.xlsx")
summary = data.reset_index()
summary['type']= summary['Attack category'].apply(lambda x: 'Normal' if x.lower() == 'normal' else 'Attack')
result = summary.groupby('type')['Number of events'].sum()
result.to_excel("attack_vs_normal_events.xlsx")
```
